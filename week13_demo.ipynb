## Part 2: NumPy Arrays vs. Python Lists

Why do we need a library like NumPy? What happens when we do math with a regular Python list?

````python
# Create a standard Python list
my_list = [10, 20, 30]

# What happens when you multiply a list?
print(f"List * 2: {my_list * 2}")
````

That is not what we wanted. The list was repeated. Now the same thing with a NumPy array.

````python
# Import the numpy library, using 'np' as the standard alias
import numpy as np

# Create a NumPy array from our list
my_array = np.array(my_list)

# What happens when you multiply a NumPy array?
print(f"NumPy Array * 2: {my_array * 2}")
````

The math was applied to every item. That is called **vectorization**. NumPy can do it because every item in an array has the same data type, usually numbers. Pandas is built on top of NumPy.

````python
# Mixing types: NumPy converts everything to one common type
mixed = np.array([1, "a", True])
print(mixed)
````

## Part 3: Loading Data with pandas

Remember in Week 8 when we opened a file, looped through it, and used `.strip()` and `.split(',')`? pandas does all of that in one command.

One detail: `pd.read_csv` assumes the first line of the file is a header row. Our demo file has one. Files without a header row need an extra argument, which we will not need today.

````python
# Import the pandas library, using 'pd' as the standard alias
import pandas as pd

# Read our CSV file into a DataFrame
df = pd.read_csv('demo_transactions.csv')

# Display the DataFrame
df
````

````python
# 3. .describe() - quick statistics for the numeric columns
df.describe()
````

## Part 5: Filtering, Grouping, and Plotting

Let's use pandas to answer a question from our data: **How much did we spend in each category?**

First, filtering. A condition inside `df[ ... ]` keeps only the rows where the condition is true.

````python
# Rows where the amount is greater than 100
df[df['amount'] > 100]
````

````python
# Step 1: keep only the 'expense' rows
expenses = df[df['transaction_type'] == 'expense']
expenses
````

Step 2: group the expense rows by category and add up each group. The result is a pandas **Series**: one column of totals, labeled by category.

````python
# Step 2: group by category and sum the amounts
category_totals = expenses.groupby('category')['amount'].sum()
print(category_totals)
````

````python
# Step 3: a bar chart of the totals
import matplotlib.pyplot as plt

category_totals.plot(kind='bar', title='Total Spending by Category')
plt.show()
````
